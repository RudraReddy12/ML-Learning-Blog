```python
import pandas as pd
import numpy as np
```

```python
from sklearn.datasets import load_breast_cancer
```

```python
data = load_breast_cancer()
data.keys()
```

```python
data.data
```

```python
data.target
```

```python
x= data.data
y = data.target
df = pd.DataFrame(x,columns=data.feature_names)
df['target'] = y
```

```python
df
```

```python
df['target'].value_counts(normalize = True)
```

```python
# To handle imbalenced data apply all the methods only on the training data
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test = train_test_split(x,y,random_state=42,test_size=0.2,stratify=y)
X_train.shape,X_test.shape,y_train.shape,y_test.shape
```

```python
pd.Series(y_train).value_counts(normalize=True)
```

```python
pd.Series(y_train).value_counts()
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
ss = StandardScaler()
ss
```

```python
X_train_transformed = ss.fit_transform(X_train)
X_train_transformed
```

```python
from imblearn.over_sampling import SMOTE
```

```python
smote = SMOTE(k_neighbors=5,sampling_strategy='auto')
```

```python
X_train_res,y_train_res = smote.fit_resample(X_train_transformed,y_train)
X_train_res.shape,y_train_res.shape
```

```python
pd.Series(y_train_res).value_counts()
```

```python
smote = SMOTE(k_neighbors=5,sampling_strategy={1:285,0:260})
```

```python
X_train_res,y_train_res = smote.fit_resample(X_train_transformed,y_train)
X_train_res.shape,y_train_res.shape
```

```python
pd.Series(y_train_res).value_counts()
```

---

```python
X_train_transformed = pd.DataFrame(X_train_res,columns=ss.get_feature_names_out())
X_train_transformed
```

```python
X_train_transformed['x28'] = X_train_transformed['x28'].apply(lambda x:'high' if x>0 else 'low')
X_train_transformed['x28']
```

```python
X_train_transformed['x29'] = X_train_transformed['x29'].apply(lambda x:'yes' if x>0 else 'no')
X_train_transformed['x29']
```

```python
X_train_transformed.head()
```

```python
from imblearn.over_sampling import SMOTENC
```

```python
smotenc = SMOTENC(categorical_features=['x28','x29'])
smotenc
```

```python
X_train_res,y_train_res = smotenc.fit_resample(X_train_transformed,y_train)
X_train_res.shape,y_train_res.shape
```

```python
X_train_res
```

- SMOTEN

```python
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/car/car.data"
columns = [
   "buying", "maint", "doors",
   "persons", "lug_boot", "safety", "target"
]


df = pd.read_csv(url, names=columns)


df.head()
```

```python
df['target'].value_counts()
```

```python
from imblearn.over_sampling import SMOTENC
```

```python
smoten = SMOTENC(sampling_strategy={'unacc':1210,'acc':600,'good':300,'vgood':'300'})
smoten
```

```python
X = df.drop('target',axis=1)
y = df['target']
X.shape,y.shape
```

```python
X_res, y_res = smoten.fit_resample(X,y)
```

- ADASYN

```python
from sklearn.preprocessing import MinMaxScaler
mms = MinMaxScaler().set_output(transform = 'pandas')
mms
```

```python
X_train_transformed = mms.fit_transform(X_train)
X_train_transformed
```

```python
from imblearn.over_sampling import ADASYN
```

```python
adasyn = ADASYN()
adasyn
```

```python
X_train_res , y_train_res = adasyn.fit_resample(X_train_transformed,y_train)
X_train_res.shape , y_train_res.shape
```

### Under Sampling

```python
# Random under sampling
from imblearn.under_sampling import RandomUnderSampler
```

```python
rus = RandomUnderSampler()
rus
```

```python
pd.Series(y_train).value_counts()
```

```python
X_train_res,y_train_res = rus.fit_resample(X_train_transformed,y_train)
X_train_res.shape,y_train_res.shape
```

```python
pd.Series(y_train_res).value_counts()
```

```python
# TOMEKLINKS
from imblearn.under_sampling import TomekLinks
```

```python
tl = TomekLinks()
tl
```

```python
pd.Series(y_train).value_counts()
```

```python
X_train_res , y_train_res = tl.fit_resample(X_train_transformed,y_train)
X_train_res.shape,y_train_res.shape
```

```python
pd.Series(y_train_res).value_counts()
```

```python
tl = TomekLinks(sampling_strategy='all')
tl
```

```python
X_train_res , y_train_res = tl.fit_resample(X_train_transformed,y_train)
X_train_res.shape,y_train_res.shape
```

```python
pd.Series(y_train_res).value_counts()
```
