```python
import numpy as np
import pandas as pd
import seaborn as sns
from sklearn.model_selection import cross_val_score
from sklearn.model_selection import StratifiedKFold
```

```python
from sklearn.datasets import load_breast_cancer
```

```python
data = load_breast_cancer()
data
```

```python
X= data.data
X
```

```python
y = data.target
y
```

```python
data.feature_names
```

```python
df = pd.DataFrame(X,columns=data.feature_names)
df['target'] = y
df.head()
```

```python
df.shape
```

```python
#No need of eda
# Assumption: all the features are having dependency with target.It is possible to predict the target from features
X = df.drop('target',axis=1)
y = df['target']
X.shape,y.shape
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=15,stratify=y)
X_train.shape,X_test.shape,y_train.shape,y_test.shape
# change the random state to check for variance
```

```python
#Assumption no feature needs preprocessing
```

```python
# simple and complex models - access the  bias and variance
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
```

```python
simple_knn = KNeighborsClassifier(n_neighbors=X_train.shape[0])
simple_knn
```

```python
simple_knn.fit(X_train,y_train)
simple_knn.score(X_train,y_train)
```

```python
y_pred = simple_knn.predict(X_test)
accuracy_score(y_pred,y_test)
```

```python
complex_knn = KNeighborsClassifier(n_neighbors=1)
complex_knn
```

```python
complex_knn.fit(X_train,y_train)
complex_knn.score(X_train,y_train)
```

```python
y_pred = complex_knn.predict(X_test)
accuracy_score(y_pred,y_test) # large scope for learning,high variance,bias low [Overfitting]
```

```python
simple_dt = DecisionTreeClassifier(max_depth=2)
simple_dt
```

```python
simple_dt.fit(X_train,y_train)
simple_dt.score(X_train,y_train)
```

```python
y_pred = simple_dt.predict(X_test)
accuracy_score(y_pred,y_test)
```

```python
complex_Dt = DecisionTreeClassifier()
complex_Dt
```

```python
complex_Dt.fit(X_train,y_train)
complex_Dt.score(X_train,y_train)
```

```python
y_pred = complex_Dt.predict(X_test)
accuracy_score(y_pred,y_test)
```

- If the model is complex tends to `Overfitting`
- If the model is simple tends to `Underfitting`

```python
# To improve aaccuracy using Hyper parameter tuning.
# using optuna
!pip install optuna
```

```python
import optuna
```

```python
#1.Objective function
#Python function
def objective(trail):
    #a. Specify Hyperparameters and its levels
        # Specify n_neighbors - range 3 to 30 
    n_neighbors = trail.suggest_int('n_neighbors',3,30)
        # Specify Weights-['uniform','distance']
    weights = trail.suggest_categorical('weights',['uniform','distance'])
    #b. Model
    knn = KNeighborsClassifier(n_neighbors = n_neighbors, weights = weights)
    #c. Metric to optimize -Optimize the validation metrics - Validation accuracy
    skf = StratifiedKFold(n_splits=5,shuffle=True)
    score = cross_val_score(knn,X_train,y_train,scoring = 'accuracy',cv = skf).mean()
    return score
```

```python
#2.Create a study
#Default sampler/search is Bayesian search - tpe
study = optuna.study.create_study(sampler=optuna.samplers.TPESampler(),study_name='knn_study',direction='maximize')
study
```

```python
#3.Optimize
study.optimize(objective,n_trials=40)
```

```python
study.best_params
```

```python
study.best_value
```

```python
# Build a model
knn = KNeighborsClassifier(**study.best_params)
knn
```

```python
knn.fit(X_train,y_train)
```

```python
knn.score(X_train,y_train)
```

```python
y_pred = knn.predict(X_test)
y_pred
```

```python
accuracy_score(y_pred,y_test)
```
