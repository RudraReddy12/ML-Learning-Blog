### Cross Validation:
- cross validation is mainly used to check how well the model performing on the unseen data(test data)
- To access the model towards genaralization,we have to build different models , during this process the model learns from test data leads to data leakage,not allowed.To address this we use cross validation.
- In cross validation we divide the training data into train and validation sets and comparing the mean train value and mean validation value if they are similar than the model os genaralized
- There are different types of Cross validations are available:
  1. k-Fold Cross-Validation
    - Data is divided into **k equal folds**
    - Model trains on **k-1 folds** and tests on the remaining fold
    - Process repeats **k times**, each fold used once as test set
    - Final score = **average of all k results**
    - Most commonly used CV method

  2. Stratified k-Fold CV
    - Same as k-fold, but **class distribution is preserved** in each fold
    - Very useful for **classification problems with imbalanced data**

  3. Leave-One-Out Cross-Validation (LOOCV)
    - Special case of k-fold where **k = number of samples**
    - Each time, **1 sample = test set**, rest = training
    - Very accurate but **computationally expensive**

  4. Leave-P-Out Cross-Validation (LPOCV)
    - Similar to LOOCV, but **P samples** are used as test set each time
    - Even more computationally costly than LOOCV

  5. Repeated k-Fold CV
    - k-Fold CV repeated **multiple times with different random splits**
    - Produces **more stable performance estimates**


```python
import numpy as np
import pandas as pd
import seaborn as sns
from sklearn.model_selection import cross_val_score
from sklearn.model_selection import StratifiedKFold
```

```python
from sklearn.datasets import load_breast_cancer
```

```python
data = load_breast_cancer()
data
```

```python
X= data.data
X
```

```python
y = data.target
y
```

```python
data.feature_names
```

```python
df = pd.DataFrame(X,columns=data.feature_names)
df['target'] = y
df.head()
```

```python
df.shape
```

```python
#No need of eda
# Assumption: all the features are having dependency with target.It is possible to predict the target from features
X = df.drop('target',axis=1)
y = df['target']
X.shape,y.shape
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=15,stratify=y)
X_train.shape,X_test.shape,y_train.shape,y_test.shape
# change the random state to check for variance
```

```python
#Assumption no feature needs preprocessing
```

```python
# simple and complex models - access the  bias and variance
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
```

```python
simple_knn = KNeighborsClassifier(n_neighbors=X_train.shape[0])
simple_knn
```

```python
simple_knn.fit(X_train,y_train)
simple_knn.score(X_train,y_train)
```

```python
y_pred = simple_knn.predict(X_test)
accuracy_score(y_pred,y_test)
```

```python
complex_knn = KNeighborsClassifier(n_neighbors=1)
complex_knn
```

```python
complex_knn.fit(X_train,y_train)
complex_knn.score(X_train,y_train)
```

```python
y_pred = complex_knn.predict(X_test)
accuracy_score(y_pred,y_test) # large scope for learning,high variance,bias low [Overfitting]
```

```python
simple_dt = DecisionTreeClassifier(max_depth=2)
simple_dt
```

```python
simple_dt.fit(X_train,y_train)
simple_dt.score(X_train,y_train)
```

```python
y_pred = simple_dt.predict(X_test)
accuracy_score(y_pred,y_test)
```

```python
complex_Dt = DecisionTreeClassifier()
complex_Dt
```

```python
complex_Dt.fit(X_train,y_train)
complex_Dt.score(X_train,y_train)
```

```python
y_pred = complex_Dt.predict(X_test)
accuracy_score(y_pred,y_test)
```

- If the model is complex tends to `Overfitting`
- If the model is simple tends to `Underfitting`

## Cross validation:Detect the genaralization

```python
from sklearn.model_selection import KFold,RepeatedKFold
```

```python
kf = KFold(n_splits=5,shuffle=True)
kf
```

```python
knn = KNeighborsClassifier()
```

```python
from sklearn.model_selection import cross_validate,cross_val_score
```

```python
score = cross_validate(knn,X_train,y_train,scoring='accuracy',cv=kf,return_train_score=True)
score
```

```python
score['train_score'].mean()
```

```python
score['test_score'].mean()
```

```python
score['test_score'].var(ddof = 0) #variance is low
```

```python
# used only to evaluate validation score
score = cross_val_score(knn,X_train,y_train,scoring='accuracy',cv=kf)
score
```

```python
score.mean()
```

```python
score.var(ddof = 0)
```
