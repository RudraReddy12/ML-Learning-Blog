# Linear Regression (Simple and Multiple)

### Correlation and Regression

A correlation is **a relationship between two variables**, `x` independent (or explanatory) variable and 'y' is the dependent (or response) variable. A **precise measure of the type and strength of a linear correlation between two variables is to calculate the correlation coefficient.**

**After verifying that the linear correlation between two variables is significant**, the next step is to determine the **equation of the line that best models the data**. This line is called a regression line, and its equation can be used to predict the value of `y` for a given value of `x`.


### What is Regression?

 **Regression establishes existence of an association relationship between a dependent variable and one or more independent variables, and not a causal relationship.**

- A regression analysis thus serves to infer or predict a further variable on the basis of one or more variables.
- For example, you might be interested in the factors that influence a person's salary, you could consider the highest level of education, the weekly working hours, and the age of a person.    

**Regression is a statistical technique**
  - used for finding the **existence of an association relationship** between a dependent variable (aka **y**, response variable or outcome variable) and an independent variable (aka **X**, explanatory variable, predictor variable or feature).
    - cannot be used for **establishing causal relationship** between two variables.

**Examples:**
- After India lost to England in a cricket test series, some people blamed the presence of players’ girlfriends and wives as the reason for the defeat and banned them from future tours. They ignored other possible reasons, like players struggling to play in English weather. **Just because two things happened together (players' families traveling and losing matches) doesn't prove one caused the other.**

- **Ice Cream and Sunburn:** In the summer, when people eat more ice cream, there are also more cases of sunburn. But eating ice cream does not cause sunburn—both happen because it’s hot and sunny outside. So, **the two things are connected (associated), but one does not cause the other**.
- **Shoe Size and Reading Ability:** If you look at shoe sizes and reading ability in children, you might find that kids with bigger feet read better. But having bigger feet doesn’t make you a better reader—both are related because as children grow older, their feet get bigger and they learn to read better. **Again, these two things are associated, but one does not cause the other.**

**More in-depth study is usually needed to determine whether there is a causal relationship between the variables.**

### Why Regression?



**Regression analysis serves two main purposes.**

1. **Measuring the effect of variables**  
   - What factors influence how well children can concentrate?  
   - Does a parent's education level or where they live affect how well their children do in school?

2. **Predicting outcomes**  
   - How long will a patient stay in the hospital?  
   - Which product is someone most likely to buy from an online store?

Regression analysis reveals how the predicted (dependent) variable changes when one of the input (independent) variables changes, allowing for both understanding influences and making predictions.

### What are the different types of Regression?

| Regression Type          | Number of Independent Variables       | Dependent Variable Level of Measurement | Independent Variable Level of Measurement  |
|-------------------------|--------------------------------------|-----------------------------------------|--------------------------------------------|
| Simple Linear Regression | One                                  | Continuous numeric              | Numeric or Categorical     |
| Multiple Linear Regression | More than one                      | Continuous numeric            | Numeric or Categorical    |
| Logistic Regression      | One or more                         | Categorical (Nominal, Ordinal)            | Numeric or Categorical                |

- **Simple Linear Regression**: Used for predicting one continuous dependent variable from one independent numeric variable.
- **Multiple Linear Regression**: Used for predicting one continuous dependent variable from multiple independent numeric variables.
- **Logistic Regression**: Used when the dependent variable is categorical (e.g., yes/no, classes), regardless of the number of independent variables.

### Key assumption in Linear Regression

- A key assumption of the simple linear regression model is that the **expected value of $y$ lies on a straight line described by the equation $w_0 + w_1 x$**, where $w_0$ and $w_1$ (read as "w zero" and "w one") are the unknown intercept and slope parameters, respectively.

- This is similar to the familiar equation of a line $y = mx + b$, where $b$ is the intercept and $m$ is the slope.


### Geometry of Line, Plane, and Hyperplane

Before exploring the mathematics of higher-dimensional spaces, it is important to acknowledge a philosophical viewpoint: **some argue that because the physical world has only three dimensions, the idea of higher dimensions is purely abstract.**
-  But, we don’t need these spaces to exist physically to be able to talk about them.

Thus, our description of higher dimensions will include precise definitions of many basic geometric concepts.
- We call $\mathbb{R}^n$ a "space" because it extends our geometric intuition from two and three dimensions to higher dimensions. Although $\mathbb{R}^n$ resembles $\mathbb{R}^2$ and $\mathbb{R}^3$, referring to an n-tuple like (5,3,2,7) as a "point" in $\mathbb{R}^4$ is only an analogy.


#### A Point

- A point in n-dimensional space is simply an ordered list of n real numbers. For example, $(5, 3, 2, 7)$ is a point in four dimensions, and $(8, 1, 3, 0, 9, 12)$ is a point in six dimensions. **Each number in the list is called a coordinate.**

- The collection of all such points with n coordinates is called $\mathbb{R}^n$, known as n-dimensional Euclidean space or simply n-dimensional space.

- It is often convenient to think of vectors and points in $\mathbb{R}^n$ as essentially the same: a vector is just the tuple of coordinates representing a point.

- Vectors in $\mathbb{R}^n$ are represented as columns of n real numbers:

$$
v = \begin{pmatrix} v_1 \\ v_2 \\ \vdots \\ v_n \end{pmatrix}
$$ where the components $v_1, v_2, \ldots, v_n$ specify the vector.

- Throughout, we use the variable $n$ to represent the number of components in a vector, which is called **the dimensionality of that vector**. This **dimensionality indicates how many independent directions or coordinates are needed to specify any point or vector** in the space $\mathbb{R}^n$.

- This abstraction allows us to generalize familiar geometric concepts from two or three dimensions to any number of dimensions.

#### A Line

- A **line** in $ \mathbb{R}^n $ is the set of points formed by starting at a fixed point and moving along a direction vector scaled by a real number.
- Formally, given a point $ \mathbf{p} \in \mathbb{R}^n $ and a direction vector $ \mathbf{v} \in \mathbb{R}^n $, any point $ \mathbf{L}(t) $ on the line is represented as:

$$
\mathbf{L}(t) = \mathbf{p} + t \mathbf{v}, \quad t \in \mathbb{R}
$$

  - Here, $ t $ is a scalar (real number) that "scales" the direction vector $ \mathbf{v} $, sweeping the point through every location along the line.

**Example**

- Suppose you start at point $ \mathbf{p} = (2, 3) $ and walk in the direction $ \mathbf{v} = (1, 1) $:

  - For $ t=0 $, you are at $ (2, 3) $ (the starting point).
  - For $ t=1 $, you move to $ (3, 4) $.
  - For $ t=-1 $, you move to $ (1, 2) $.

This set of points creates a straight line.

image.png

**Equation of a Line in 2D (Slope-Intercept Form)**

- In two dimensions, this line can also be described by the equation:

$$
y = mx + c
$$

  - $ m $ is the **slope**, telling us how steep the line is and $ c $ is the **y-intercept**, where the line crosses the y-axis.

- For example, the line $ y = 2x + 1 $ means that if $ x=0 $, $ y=1 $, and for every increase of 1 in $ x $, $ y $ increases by 2.

- In machine learning, the concept of a line generalizes to higher dimensions as a **hyperplane**:

$$
w^\top x + b = 0
$$

- Here, $ w $ is a weight vector (directional coefficients), $ x $ is the input vector (data point) and $ b $ is the bias or offset.

#### A Plane



- A **plane** in $ \mathbb{R}^3 $ is determined by a point $ \mathbf{p} = (a, b, c) $ on the plane and two direction vectors $ \mathbf{v} $ and $ \mathbf{u} $ that lie **parallel** to the plane.

- The fact that a plane requires **two direction vectors** (instead of one for a line) reflects that the plane is **two-dimensional**, while a line is **one-dimensional**.

- The plane consists of all points $ \mathbf{x} = (x, y, z) $ such that the displacement vector from $ \mathbf{p} $ to $ \mathbf{x} $ can be expressed as a linear combination of the two direction vectors:

$$
\mathbf{x} = \mathbf{p} + t \mathbf{v} + s \mathbf{u}, \quad t, s \in \mathbb{R}
$$

- This is called the **parametric equation of the plane** where the scalars $ t $ and $ s $ "scale" the direction vectors $ \mathbf{v} $ and $ \mathbf{u} $, sweeping out every point on the plane.

- Just like with lines, multiple choices of the direction vectors $ \mathbf{v} $ and $ \mathbf{u} $ can yield different parametric forms but describe the **same plane**, as long as the vectors lie parallel to the plane.


- An alternative way to describe a plane uses the **normal vector** $ \mathbf{n} $, which is **perpendicular (orthogonal) to the plane**.

- Given $ \mathbf{p} \in \mathbb{R}^3 $ on the plane and a normal vector $ \mathbf{n} = (a, b, c) $, all points $ \mathbf{x} = (x, y, z) $ on the plane satisfy:

$$
\mathbf{n} \cdot (\mathbf{x} - \mathbf{p}) = 0
$$

or equivalently:

$$
a(x - a) + b(y - b) + c(z - c) = 0
$$

- This is called the **normal form** of the plane equation.

#### Hyperplane



A hyperplane is an **(n-1)-dimensional subspace** in an n-dimensional space that divides the space into two distinct regions.

In the geometry of machine learning models, the equation $$w^\top x + b = 0 $$ defines a **line** (2D), **plane** (3D), or more generally, a **hyperplane** (n-dimensions).

Where:

  - **w** is called the **normal vector**, meaning it is perpendicular to the hyperplane (or to the line/plane, in lower dimensions).
  - **w** = (w1, w2, ..., wn) is the weight vector, which determines the orientation of the hyperplane.
  - **x** = (x1, x2, ..., xn) is the feature vector representing data points.
  - **b** is the bias term, which determines the position of the hyperplane.

- In 2D, **w** is perpendicular to the separating line; in 3D, it is perpendicular to the plane; in higher dimensions, it is perpendicular to the hyperplane.

  - **In 2D space:** A hyperplane is simply a line that divides the plane into two parts.

  - **In 3D space:** A hyperplane is a plane that divides the space into two regions.

  - **In higher dimensions:** A hyperplane remains an abstract construct that still serves as a decision boundary.




**Why is w Perpendicular?**

- Understanding **w** as the normal vector helps visualize and generalize classification boundaries, even in spaces beyond our ability to picture.

- Suppose you are in three dimensions. If you have a normal vector $ w = (a, b, c) $ and a point $ P_0 = (x_0, y_0, z_0) $ on a plane, any other point $ P = (x, y, z) $ on the plane satisfies:

$$
w \cdot (P - P_0) = 0
$$

- This is because the vector from $ P_0 $ to $ P $ lies in the plane (it’s parallel to the plane), and **w** is defined to be perpendicular. Expanding this:

$$
a(x - x_0) + b(y - y_0) + c(z - z_0) = 0
$$

- Rearranging gives the general equation for a plane:

$$
ax + by + cz + d = 0
$$

- ($d$ collects constants). The coefficients $ a, b, c $ are the components of the normal vector.

- This principle holds in any number of dimensions. The following equation defines a hyperplane, and the vector **w** is perpendicular to it.

- In machine learning, this is how classifiers decide the boundary that separates categories—by computing on which side of the hyperplane, plane, or line a new data point falls.

$$
w^\top x + b = 0
$$

- If $ w^\top x + b > 0 $, you’re on one side; if less than zero, you’re on the other; if zero, the point lies exactly on the boundary.



| Geometry      | Equation Form           | Dimension | Perpendicular Vector      |
|---------------|------------------------|-----------|--------------------------|
| Line          | $$ w_1 x_1 + w_2 x_2 + b = 0 $$ | 2         | $$ (w_1, w_2) $$         |
| Plane         | $$ w_1 x_1 + w_2 x_2 + w_3 x_3 + b = 0 $$ | 3         | $$ (w_1, w_2, w_3) $$    |
| Hyperplane    | $$ w^\top x + b = 0 $$            | n         | $ w $ (n components)   |

#### Parallel vectors

- A **vector** $ \mathbf{v} $ is said to be **parallel to a plane** if it can be positioned so that it lies entirely within the plane.

- Equivalently, $ \mathbf{v} $ is parallel to a plane if there exist two points $ \mathbf{p} $ and $ \mathbf{q} $ on the plane such that:

$$
\mathbf{v} = \mathbf{q} - \mathbf{p}
$$

- Any vector parallel to the plane is **orthogonal to the plane's normal vector** $ \mathbf{n} $.

- Conversely, any vector orthogonal to the normal vector is parallel to the plane.



**Example:**  
- For the plane $ 2x + 4y + 5z = 12 $, the normal vector is $ \mathbf{n} = (2,4,5) $. The vector $ (3, 1, t) $ is parallel to the plane if:

$$
(2,4,5) \cdot (3,1,t) = 0 \implies 6 + 4 + 5t = 0 \implies t = -2
$$

- The **cross product** of any two nonparallel vectors in the plane produces a normal vector perpendicular to both and thus perpendicular to the plane itself.

#### Distance from a point to a plane

Given a plane in $\mathbb{R}^3$ defined by

$$
a x + b y + c z + d = 0,
$$

and a point $\mathbf{p} = (x_0, y_0, z_0)$ not on the plane, there is exactly one point $\mathbf{q}$ on the plane closest to $\mathbf{p}$. This point $\mathbf{q}$ is the **projection** of $\mathbf{p}$ onto the plane.

- The **distance** from $\mathbf{p}$ to the plane is the length of the perpendicular dropped from $\mathbf{p}$ to the plane.

- Using the dot product and unit normal vector

$$\mathbf{u} = \frac{(a, b, c)}{\sqrt{a^2 + b^2 + c^2}}$$, the distance $d$ can be calculated as:

$$
d = \frac{|a x_0 + b y_0 + c z_0 + d|}{\sqrt{a^2 + b^2 + c^2}}
$$

In the boundary-centric view of machine learning, the **dot product** plays a foundational role in both regression and classification:

- For **regression**, the prediction hyperplane is modeled as the dot product of weights and input variables, plus a bias term: $$ y = w^\top x + b $$. This is a direct application of the dot product, which acts as a “projection” of the inputs onto the weight vector, yielding a scalar output for the regression estimate.
- In **classification** tasks, the class assignment for a point is determined by the distance from the point to the decision boundary. This is computed as the dot product of the feature vector and the weight vector (plus bias), with the sign and magnitude indicating on which side of the boundary the point lies (vector lands relative to the decision boundary)—mirroring the "projection" interpretation of the dot product.
- In **SVMs (Support Vector Machines)**, the margin and the decision function are defined via the dot product between the data point and the hyperplane’s normal vector.  
- **Neural networks** also rely on the dot product to compute neuron activations, as each node calculates a weighted sum (dot product) of its inputs—a key part of forward propagation.

The dot product operation is often applied in data science and machine learning… Neural networks use dot products to compute weighted sums efficiently".

### Ordinary Least Squares (OLS)



The equation of the best fit regression line $ Y = \beta_0 + \beta_1 X $ is found by minimizing the cost function, specifically the Residual Sum of Squares (RSS), using Ordinary Least Squares (OLS). There are two common approaches to do this:

1. **Differentiation (Analytical Method):**  
   Take the derivative of the RSS cost function with respect to $\beta_0$ and $\beta_1$, set them to zero, and solve the resulting equations to get the exact solution (known as the Normal Equation).

2. **Gradient Descent (Iterative Method):**  
   Instead of solving directly, start with initial estimates of $\beta_0$ and $\beta_1$, then iteratively update them by moving in the direction that reduces the cost function until convergence.


- In linear regression, the goal is to find the best-fitting line that models the relationship between input variables (features) and an output variable (target).

  - The model is represented by a linear function where the prediction is a dot product of weights and input features plus a bias term:

$$
\hat{y} = w^\top x + b
$$

  - Best fitting line means minimize the sum of squared errors (residuals)

#### **Analytical (Closed-Form) Solution — Differentiation**





- To quantify the quality of this fit, an **objective function** (commonly the Mean Squared Error, MSE) measures the average squared difference between predicted values $\hat{y}_i$ and actual values $y_i$:

$$
J(w, b) = \frac{1}{n} \sum_{i=1}^{n} (\hat{y}_i - y_i)^2
$$

- The optimization process involves minimizing this objective function by adjusting the model parameters (weights and bias).

- The simplest idea is to evaluate the function at **all possible values of w and b** to find the optimum.
- This gives the **Normal Equation:**

$$
w^* = (X^T X)^{-1} X^T y
$$

- **Pros:**
  - Exact solution without approximation.
  - Efficient for small to medium datasets with moderate features.
- **Cons:**
  - Requires matrix inversion, which is costly for large datasets.
  - Not suitable if $X^TX$ is singular or data is very high dimensional.
  - Evaluations must be finite, making exhaustive search impossible for continuous domains.
  - Even with a discrete grid of points, the number of evaluations grows exponentially with problem size, causing extreme computational demands.\
  - Another approach is to use **inequalities to establish upper or lower bounds** on the function’s value, then attempt to find if these bounds are attainable.

    - This method is often infeasible for complex or non-convex functions, such as neural network loss functions.

- Thus, **efficient optimization methods** are necessary, which iteratively guide search towards minimum or maximum values without exhaustively sampling points.



#### Why Directly Solving $ \frac{d}{dw}L(w) = 0 $ Often Fails in Practice???

- **Must Satisfy two conditions** 1. “Tall Matrix” Requirement and 2. Linear Independence (No Perfect Multicollinearity)

- **Multiple types of stationary points** exist in a derivative equation: minima, maxima, and saddle (inflection) points. Simply solving the derivative equal zero doesn’t guarantee a minimum.

- **Analytical solution is only feasible for simple problems.** For example, linear regression with mean squared error has a closed-form solution via the normal equation.

- **For complex models like neural networks,** the loss functions are highly nonlinear and non-convex, making closed-form solutions unavailable or computationally prohibitive.

- **Iterative optimization methods (e.g., gradient descent)** are used instead to progressively move towards minima by following the gradient.

#### **Gradient Descent — Iterative Numerical Optimization**



- Start with initial parameters and **iteratively update** them to minimize the cost function.
- Suppose we want to fit a simple linear model $ y = wx $ that minimizes the loss function

$$
L(w) = \frac{1}{n} \sum_{i=1}^n (wx_i - t_i)^2
$$

where $ t_i $ are true labels.

The gradient of $ L(w) $ w.r.t. $ w $ is

$$
\frac{dL}{dw} = \frac{2}{n} \sum_{i=1}^n x_i (wx_i - t_i)
$$

Instead of solving $ \frac{dL}{dw} = 0 $ analytically, we update $ w $ iteratively using

$$
w \leftarrow w - \alpha \frac{dL}{dw}
$$

where $ \alpha $ is the learning rate.

To understand gradient descent, consider another viewpoint: derivatives as vectors. For any differentiable function $ f(x) $, its derivative $ f'(x) $ can be interpreted as a one-dimensional vector.
- If $ f'(x) $ is positive,
  - the vector points to the right;
- Else (i.e., If $ f'(x) $ negative),
  - it points to the left.


- **Pros:**
  - Scales well to large datasets and high-dimensional features.
  - Flexible and works with various differentiable cost functions.
  - Core method for training neural networks and complex ML models.
- **Cons:**
  - Requires tuning of the learning rate.
  - May converge slowly or get stuck in local minima for non-convex problems.

**Summary Table**

| Method           | How it Works                      | Pros                                    | Cons                              |
|------------------|---------------------------------|----------------------------------------|----------------------------------|
| Differentiation   | Solve derivative = 0 (Normal Eq)| Exact; good for small problems         | Expensive for large datasets; poor scaling   |
| Gradient Descent  | Iterative parameter updates     | Scales well to big data and complex models | Requires tuning; approximate; slower convergence|

**Rule of Thumb**
- Use **Normal Equation** for small/medium datasets with fewer than ~10,000 features.
- Use **Gradient Descent** for large datasets, high-dimensional data, or when regularization is involved.

````python
import numpy as np
````

````python
def func(x):
  return x**2
def grad(x):
  return 2*x
````

````python
x = np.linspace(-3, 3, 100)
````

````python
import seaborn as sns
````

````python
sns.lineplot(x = x, y = func(x))
````

````python
lst = []
x_old = -3
learn_rate = 0.05
for _ in range(100000):
  x_new = x_old - learn_rate*(grad(x_old))
  lst.append((x_old, func(x_old)))
  # print(x_old, square(x_old))
  x_old = x_new
````

````python
import pandas as pd
````

````python
df = pd.DataFrame(lst)
sns.lineplot(x = x, y = func(x))
sns.scatterplot(x = df[0], y = df[1], color = 'red')
````

## Model assumptions and common violations

In simple terms, the ordinary least squares (OLS) method in linear regression works best when certain model assumptions are true. These are the rules about the data and errors that help ensure the estimates of the model's coefficients (the β values) are reliable. Specifically, under these assumptions:

- The OLS estimates are **unbiased**, meaning on average, the estimated coefficients equal the true coefficients.
- They have the **smallest possible variance** among all linear unbiased estimators, which means the estimates are as precise as possible.

However, if one or more of these assumptions are violated, the OLS estimates may no longer be reliable. Not only can the coefficient estimates become biased or less precise, but statistical tests (like significance tests) that depend on these assumptions can also become invalid. Understanding and checking these assumptions is crucial before trusting the regression results.

#### How to check the assumptions of OLS model?

**Residual plots** are a simple but powerful tool to check whether the assumptions of a linear regression model hold true.
- Residuals are the difference between **observed values** and the values **predicted by the model**.
- Residuals are calculated as $e = y - \hat{y}$, the difference between **actual and predicted values**.  
- By plotting residuals against predicted values or individual features, we can visually detect if the model fits well.

  - When **residuals are randomly scattered around zero** without a clear pattern, it indicates that **model assumptions like linearity and constant error variance hold**.  
  - **Patterns or trends in residual plots** suggest violations of these assumptions and that the **model may be inadequate**.  
- Residual plots help identify outliers, which stand out with unusually large residuals. Outliers may need correction or removal if due to data errors.  
- Residual plots serve as informal diagnostics to guide improving the model before formal statistical tests.  

#### Linear and additivity of the relationship between dependent ($y$) and independent variables ($X$).

For a linear regression model to be valid, the independent variable(s) $X$ and the dependent variable $y$ should exhibit some form of linear association.

- Without a linear relationship, fitting a linear model is pointless as it will fail to capture the true connection.
  
- **How to detect it?**  
  A scatter plot showing points closely aligned along a straight line or gently sloping curve indicates linearity. In contrast, a random scatter or clear nonlinear shape warns against linear modeling.

- **How to fix it?**
  - Apply appropriate nonlinear transformations to response or predictor variables (e.g., log or power transformations). Log transformations are suitable for strictly positive data and can convert multiplicative relationships into additive ones.
  - For suspected polynomial relationships, include terms like $X^2$ or higher powers as additional predictors.
  - Avoid adding arbitrary complexity; unnecessary polynomial terms may cause overfitting and unreliable extrapolation.
  Consider introducing interaction terms (e.g., $X_1 X_2$) or piecewise variables when relationships depend on other predictors or change over time.

#### Error Terms are Normally Distributed with Mean Zero (Errors, Not $X$, or $y$)

The residuals or errors $$\varepsilon = y - \hat{y}$$ — differences between observed and predicted values — should ideally be normally distributed around zero.

-  Normality of errors ensures reliable inference, especially hypothesis tests and confidence intervals for coefficients. If errors deviate substantially from normality, p-values and related tests may be invalid.
- If you're interested only in predicting the line and not inference, strict normality is less critical.

- **How to detect it?**
  A histogram or a Q-Q plot of residuals showing the classic bell-shaped curve confirms normality. Deviations or skewness indicate violation.
- **How to fix it?**
   - Apply nonlinear transformations to the dependent variable (e.g., logarithm, square root) to reduce skewness and stabilize variance, making residuals more normally distributed.
   - Use methods less sensitive to outliers and departures from normality, such as least absolute deviations regression or M-estimators.
   - Identify influential outliers via diagnostic plots and consider removing, correcting, or modeling them separately if justified.



#### Error Terms are Independent

Each error term should be independent of others, especially important when data involves time or sequence (e.g., time series).

- Dependence (autocorrelation) can bias estimates and standard errors, affecting model validity.

- **How to detect it?**
  A plot of residual values over time with no obvious pattern signifies independence. Patterns such as consecutive runs of positive or negative residuals suggest dependence.

#### Error Terms Have Constant Variance (Homoscedasticity)

The variance of errors should remain stable regardless of the level of the independent variable(s).

- Changing error variance (heteroscedasticity) distorts regression standard errors and invalidates tests.

- **How to detect it?**
  A residuals vs. fitted values plot showing a consistent scatter without a funnel or pattern shape indicates homoscedasticity. Spread increasing or decreasing with fitted values signals heteroscedasticity.

- **How to fix it?**
  - Apply nonlinear transformations (e.g., logarithm) to dependent and/or independent variables, especially when data are positive and errors scale with prediction magnitude.  
  - In time series, adjust for inflation or compound growth via logging or deflating.  
  - For financial data with volatile periods, consider segmenting data or using specialized models like ARCH (autoregressive conditional heteroscedasticity).  
  - Add polynomial or interaction terms or include seasonal dummy variables to model seasonal heteroscedastic patterns carefully.


| Assumption                      | Detection                          | Description                                          |
|--------------------------------|---------------------------------------|------------------------------------------------------|
| Linear Relationship             | Scatter plot                          | Points aligned roughly along a straight line         |
| Normality of Errors             | Histogram / Q-Q plot                  | Residuals bell-shaped or lie on Q-Q plot line        |
| Independence of Errors          | Residuals over time plot              | Random scatter with no temporal structure            |
| Constant Variance (Homoscedasticity) | Residuals vs. fitted values scatter plot | Uniform spread of residuals across predicted values  |


#### There should be no exact linear relationship (no perfect multicollinearity) among explanatory variables in a regression model.


- **Perfect multicollinearity occurs if some variables can be exactly expressed as a linear combination of others, making model estimation impossible.** For example, if the number of cars sold (x1) and cars remaining in stock (x2) add up to a constant, these are perfectly collinear.

- **Non-perfect multicollinearity, where variables are highly correlated but not exact linear combinations, is common**. It leads to imprecise coefficient estimates, making it hard to separate the individual effect of predictor variables. Severe multicollinearity can cause important variables to appear insignificant or have incorrect coefficient signs.

**Detecting Multicollinearity**  
- Look for high correlation (>0.8 or <-0.8) between predictors.  
- A high overall $R^2$ but insignificant individual coefficients suggests multicollinearity.  
- Use Variance Inflation Factor (VIF); values > 5 or 10 indicate multicollinearity.

**Remedies**  
- Drop one of the correlated variables.  
- Collect more data to reduce correlation.  
- Transform variables to reduce collinearity.  
- In some cases, all variables are justified and can be kept.

## Transformation

- In classical statistics and machine learning, **linear models** are powerful tools when a *linear* relationship exists between a predictor variable $X$ and a target variable $Y$. The modeling process involves identifying optimal parameters so the model best predicts $Y$ from $X$.

- However, real-world data often display nonlinear relationships. Fitting a linear model to such data can lead to poor performance and misleading conclusions. **Fortunately, through transformation of the data, we can often *linearize* these relationships—making linear analysis both feasible and effective**.

Transformations serve to **straighten** curved relationships between variables so they better conform to the assumptions of linear regression. Common transformations include:

- Taking the log, square root, reciprocal, or exponentiating variables.
- Raising variables to different powers.



#### The Tukey-Mosteller Bulge Diagram

John Tukey and Frederick Mosteller introduced the **Bulge Diagram (or Rule of the Bulge)** as a graphical reference for choosing effective variable transformations.

- **How it works:**  
  - Plot $X$ versus $Y$. Observe the “bulge” or curvature in the scatterplot.
  - Compare the plot to the four quadrants of the Bulge Diagram, each representing a broad type of nonlinearity.
  - Each quadrant suggests power transformations for $X$, $Y$, or both to straighten the relationship. The diagram depicts which transformations (e.g., log, power, reciprocal) will “shrink” or “expand” certain parts of the curve, helping to achieve linearity.



image.png

- **Key Points:**  
   - **Upward/Downward transformations:**  
     - Square-root and logarithm make values “smaller” (down the “ladder of powers”).  
     - Raising to a higher power makes values “bigger”.

   - **Multiple Solutions:**  
     - More than one transformation may improve linearity; experimentation and interpretation are encouraged.
   - **Additional Goals:**  
     - Besides linearity, transformations can improve symmetry, stabilize variance, or meet model diagnostic needs.

- **Example:**  
  - If a scatter plot of $X, Y$ shows a curve resembling Quadrant 2 of the Bulge Diagram, you might try a log transformation on $X$ and a square-root on $Y$ to achieve linearity.

## Tests of Significance: Is the beta coefficient/s significant?



The fundamental question is: **Is the beta coefficient significant?**



Significance testing using hypothesis tests (commonly a t-test) answers this question, guiding whether a regression model’s slope should be considered meaningful or ignored in practice

- Tests of significance in linear regression help us decide if the explanatory variables have a meaningful relationship with the response variable by testing the unknown regression coefficients $\beta_0, \beta_1, ..., \beta_k$.

- Before building a linear regression model, it’s important to address one essential theoretical question—the significance of the beta coefficient, specifically the slope ($\beta_1$). Fitting a regression line gives you the intercept ($\beta_0$) and slope ($\beta_1$), but it’s crucial to determine if $\beta_1$ meaningfully represents a real relationship between the predictor and response variable.

- This means testing whether the estimated slope truly reflects a linear relationship between $X$ and $y$ and is not just a result of random chance in the data.
  - A significant beta coefficient suggests that changes in the input variable are associated with systematic changes in the target variable.
  - If it’s not significant, the line fitted by the model may not be useful for prediction or explanation, regardless of its calculated values.



**Test of Joint Significance**

Checks if **all** slope coefficients are zero ($H_0: \beta_1 = \beta_2 = \cdots = \beta_k = 0$) versus at least one is nonzero ($H_A$).
- This test evaluates if the model as a whole explains variation in the response variable.
- The test uses an **F-statistic** which compares explained variance (MSR) to unexplained variance (MSE).

- It uses an **F-statistic**, which compares the model’s explained variance (Mean Square Regression, MSR) against unexplained variance (Mean Square Error, MSE):

$$
F = \frac{MSR}{MSE} = \frac{SSR/k}{SSE/(n-k-1)}
$$

where $SSR$ is the regression sum of squares, $SSE$ is the error sum of squares, $k$ is the number of predictors, and $n$ is the sample size.

- A **low p-value** for the F-test means rejecting $H_0$, indicating the model as a whole is statistically significant and useful.
- Conversely, if the p-value is high, it suggests that none of the explanatory variables jointly have an effect on the response.


**Test of Individual Significance**

Tests whether a single coefficient $\beta_j$ differs from a specified value, typically zero.

- Hypotheses:  
  - Two-tailed: $H_0: \beta_j = 0$ vs $H_A: \beta_j \neq 0$  
  - One-tailed for direction if prior knowledge exists.
- Uses a **t-statistic**, which is the coefficient estimate divided by its standard error.

- This test uses a **t-statistic** calculated as:

$$
t = \frac{\hat{\beta}_j - 0}{SE(\hat{\beta}_j)}
$$

where $\hat{\beta}_j$ is the estimated coefficient and $SE(\hat{\beta}_j)$ its standard error.

- Rejecting $H_0$ means the corresponding variable significantly affects the response.

**After fitting an OLS regression model, the key values to examine are:**

- **F-statistic and its p-value:**  
  The F-statistic tests whether the overall regression model provides a significant fit to the data, considering all predictors together. A low p-value for the F-statistic (typically < 0.05) indicates that the model as a whole is significant and not just a result of random chance.

- **R-squared:**  
  This value shows the proportion of variance in the response variable that is explained by the model. Higher R-squared values indicate a better fit.

- **Coefficients and their p-values:**  
  For each predictor, the coefficient p-value assesses whether that variable significantly contributes to the prediction of the outcome. Significant coefficients (p-value < 0.05) suggest that the corresponding input feature has a meaningful impact on the response.

These metrics help determine both the overall usefulness of the model and the individual importance of each predictor.

**Simple Linear Regression: Moore-Penrose Pseudoinverse Solution**

**Data**: $ n $ observations $(x_i, y_i)$, $ i = 1, 2, ..., n $

**Design Matrix** (with intercept):
$$
\mathbf{X} = \begin{bmatrix}
1 & x_1 \\
1 & x_2 \\
\vdots & \vdots \\
1 & x_n
\end{bmatrix}, \quad
\mathbf{y} = \begin{bmatrix}
y_1 \\
y_2 \\
\vdots \\
y_n
\end{bmatrix}
$$

**Model**:

$$
\mathbf{y} = \mathbf{X}\boldsymbol{\beta} + \boldsymbol{\epsilon}
$$

where $\boldsymbol{\beta} = [\beta_0, \beta_1]^T$

**Moore-Penrose Solution**

**Optimal coefficients**:
$$
\hat{\boldsymbol{\beta}} = (\mathbf{X}^T \mathbf{X})^{-1} \mathbf{X}^T \mathbf{y} = \mathbf{X}^+ \mathbf{y}
$$

Where $\mathbf{X}^+$ is the **Moore-Penrose pseudoinverse**.

**Step-by-Step Solution**

```
1. Construct design matrix X (n×2): column 1 = 1's, column 2 = x values
2. Compute XᵀX (2×2 matrix)
3. Invert XᵀX → (XᵀX)⁻¹ (2×2)
4. Compute Xᵀy (2×1 vector)
5. Multiply: β̂ = (XᵀX)⁻¹ Xᵀy
```

````python
# Squared Error
def se(X, y, w, b):
  y_hat = w*X.squeeze() + b
  sqerr = ((y - y_hat)**2)
  ssqer = sqerr.sum()
  return ssqer
````

````python
# Gradient - 2*(y - y_pred)*(-w*X)
def grad_w(X, y, w, b):
  y_pred = w*X.squeeze() + b
  grad = -2*sum(X.squeeze()*(y - y_pred))
  return grad
````

````python
# Gradient - 2*(y - y_pred)*(-1)
def grad_b(X, y, w, b):
  y_pred = w*X.squeeze() + b
  grad = -2*sum((y - y_pred))
  return grad
````

````python
# Gradient Descent
# w_new = w_old - eta*(grad_w/b)
w_hist = []
w, b = 0, 0
eta = 0.00008
n_iter = 1000
iter = 0
while iter < n_iter:
  error = se(X_train_transformed, y_train, w, b)
  gw = grad_w(X_train_transformed, y_train, w, b)
  gb = grad_b(X_train_transformed, y_train, w, b)
  w = w - (eta*gw)
  b = b - (eta*gb)
  error_new = se(X_train_transformed, y_train, w, b)
  w_hist.append((w, b, error_new))
  iter += 1
  # Plot
  if iter%10 == 0:
    y_hat = w*X_train_transformed.squeeze() + b
    sns.scatterplot(x = X_train_transformed.squeeze(), y = y_train)
    sns.lineplot(x = X_train_transformed.squeeze(), y = y_hat, color = 'r')
    plt.title(f'w: {w: .2f}, b: {b:.2f}, error: {error_new:.2f}')
    plt.show()
  # if error_new > error:
  #   iter = n_iter
````

## Multiple Linear Regression

Multiple linear regression extends the concept of simple linear regression by allowing multiple predictor variables to explain or predict a response variable. The model predicts the response using the formula:

$$
Y = \beta_0 + \beta_1 X_1 + \beta_2 X_2 + \ldots + \beta_p X_p + \epsilon
$$

where:
- $ Y $ is the dependent variable,
- $ X_1, X_2, \ldots, X_p $ are the independent predictor variables,
- $ \beta_0 $ is the intercept (expected value of $Y$ when all $X$'s are zero),
- $ \beta_1, \ldots, \beta_p $ are the slope coefficients representing the effect of each predictor,
- $ \epsilon $ is the error term.


Each coefficient $ \beta_j $ represents the **expected change in the mean response $ E(Y) $** associated with a one-unit increase in $ X_j $, **holding all other predictors constant**. This interpretation adjusts for the influence of multiple variables simultaneously, reflecting the unique contribution of each predictor.

- Instead of fitting a straight line (as in simple linear regression), here the model fits a **hyperplane** in the multi-dimensional predictor space.
- Coefficients are still estimated by minimizing the **sum of squared errors** (least squares criterion).
- The **assumptions** underlying inference remain the same: errors should be independent, normally distributed with mean zero, and have constant variance (homoscedasticity).
- **Normalization or Standardization** is a technique that not only performs mean centering but also feature scaling. In multivariate regression, **if
the scales of xi and xj are significantly different, this can also affect the estimate of w and b, so we need to standardize the scales of all x**.


#### New features: New considerations

###### **1. Overfitting in Multiple Linear Regression**



As you add more variables to a multiple linear regression model, the model can become overly complex. An overfitted model memorizes the training data instead of learning general patterns, leading to excellent performance on training data but poor generalization to new, unseen data. Overfitting is typically evident when the model shows high accuracy on training data but very low accuracy on test data.

**How to Detect Overfitting:**  
- Compare model performance on training vs. validation/test sets.  
- Use techniques like cross-validation or predicted R-squared to assess model generalizability.  
- A large gap between training and test accuracy indicates overfitting.

**How to Fix Overfitting:**  
- Increase the size and quality of training data.  
- Use feature selection to remove irrelevant or redundant predictors.  
- Apply regularization techniques (like Ridge, Lasso) to penalize model complexity.  
- Simplify the model by reducing features or degrees of freedom.

##### **2. Multicollinearity in Multiple Linear Regression**

Multicollinearity occurs when predictor variables are highly correlated with each other, making it difficult to isolate each variable’s unique effect on the response. This inflates the variance of coefficient estimates, leading to unstable and unreliable estimates.

This relationship causes redundancy, meaning that the inclusion of some variables does not add new information to the model. Multicollinearity complicates interpretation because it undermines the assumption that each coefficient represents the change in the response variable while holding other variables constant.

**Effects of Multicollinearity:**

- **Interpretation becomes unclear:** The idea of measuring the effect of one predictor while keeping others constant fails because predictors move together.
- **Unstable coefficients:** The estimated coefficients can vary wildly depending on the sample data, sometimes reversing signs unexpectedly.
- **Unreliable inference:** P-values for coefficients lose reliability, making hypothesis testing questionable.

**Despite these issues, multicollinearity does not affect:**
- The model’s ability to make predictions.
- Goodness-of-fit statistics such as R-squared.

**How to Detect Multicollinearity:**
- Same as detecting associations in predictors. Scatter plots are used to visually inspect it and Correlations coefficients are also used to quantify the linear association.
  - However, pairwise correlations alone might not capture the full picture. Sometimes, **an independent variable can be highly correlated not just with one other variable, but with a combination of several other variables simultaneously.** In such cases, pairwise analysis isn’t sufficient.

  - The idea is to **construct a model that predicts one independent variable based on the other independent variables. This helps assess how much the variable can be explained or predicted by the combination of others. It is a useful approach to detect multicollinearity because it shows if an independent variable depends strongly on the other predictors collectively**—not just on them individually.

- One of the most popular ways to detect multicollinearity is using the **Variance Inflation Factor (VIF)**. VIF quantifies how much the variance of a regression coefficient is inflated due to correlation with other predictors.

- Calculate the Variance Inflation Factor (VIF) for predictors; high VIF values (commonly above 5 or 10) suggest multicollinearity.
  - A **VIF of 1** indicates no correlation.
  - A **VIF between 1 and 5** suggests moderate correlation, usually acceptable.
  - A **VIF above 5 (or 10)** indicates high multicollinearity and merits investigation.
- Examine correlation matrices for high pairwise correlations among predictors.

**How to Fix Multicollinearity:**  
- **Remove or combine correlated features:** Drop variables with high VIF or combine related variables using techniques like Principal Component Analysis (PCA).  
- **Add interaction features**: Create new variable using the interactions of the older variables. Add interaction features, drop original features
- **Regularization methods:** Use Ridge or Lasso regression, which penalize large coefficients and mitigate multicollinearity effects.  
- **Feature engineering:** Create new features or interaction terms to reduce dependence among variables.

###### **What is Variance Inflation Factor (VIF)?**



- The term `inflation` in Variance Inflation Factor (VIF) refers to the increase or magnification of the variance of an estimated regression coefficient caused by multicollinearity among predictors.

- In multiple linear regression, each estimated coefficient has an associated variance that reflects the uncertainty or precision of that estimate. **When predictor variables are correlated (multicollinearity), this variance increases—meaning the uncertainty about the coefficient grows.**
- The VIF quantifies how much larger the variance of a coefficient is compared to what it would be if the predictor were uncorrelated with all other predictors. Essentially, **VIF tells us the factor by which the variance (or standard error squared) of the coefficient is “inflated” because of the linear relationships among predictors.**

- In a multiple regression model, the standard errors—and thus the variances—of estimated coefficients can inflate when predictors are correlated. The VIF for a particular predictor $X_j$, denoted as $ \text{VIF}_j $, measures this inflation. It answers: by what factor is the variance of the coefficient $ \hat{\beta}_j $ inflated due to the correlation between $ X_j $ and the other predictors?

Mathematically, the VIF for predictor $ X_j $ is:

$$
\text{VIF}_j = \frac{1}{1 - R_j^2}
$$

Where $ R_j^2 $ is the coefficient of determination obtained by regressing $ X_j $ on all the other predictor variables in the model.

**Step-by-Step Explanation**
1. Regress predictor on others
  Take variable $X_j$. Try to predict it using all the other predictors $X_1, X_2, \dots, X_{j-1}, X_{j+1}, \dots, X_p$.
  This gives you $R_j^2$, the amount of variance in $X_j$ explained by the other predictors.

2. Interpret $R_j^2$
  - If $R_j^2$ is low → $X_j$ is independent of others → little redundancy.
  - If $R_j^2$ is high → $X_j$ is highly correlated with others → lots of redundancy.
3. Plug into VIF
$$\text{VIF}_j = \frac{1}{1 - R_j^2}$$

  - If $R_j^2 = 0 → \text{VIF}_j = 1 →$ no inflation.
  - If $R_j^2 = 0.9 → \text{VIF}_j = \frac{1}{0.1} = 10 →$ variance of coefficient is inflated 10 times compared to uncorrelated predictors.
4. Meaning of VIF Values
  - VIF = 1 → No multicollinearity (predictor independent).
  - VIF = 1–5 → Moderate correlation, usually acceptable.
  - VIF > 10 → High multicollinearity; coefficients are unstable and unreliable.

##### **3. Feature Selection in Multiple Linear Regression**



With many potential predictors, selecting an optimal subset that best explains the response variable is crucial. Feature selection reduces redundancy and improves model interpretability and performance.

**How to Perform Feature Selection:**  
Searching through *all* possible feature combinations (exhaustive search) to select an optimal regression model quickly becomes impractical, since for $p$ predictors, $2^p$ models must be checked—an approach that’s computationally infeasible for even moderately sized datasets.

- **Manual feature elimination** offers a practical alternative:
  - Start by including all available predictors in the initial model.
  - Remove predictors that show little statistical evidence of usefulness (high p-values).
  - Drop predictors that are redundant, as revealed through high pairwise correlations and elevated Variance Inflation Factors (VIF).
  - Refit and iterate, evaluating model fit and parsimony at each step.

- **Automated methods** have emerged from research to streamline feature selection and mitigate limitations of manual tuning:
  - **Recursive Feature Elimination (RFE):** Ranks predictors by their importance and removes least-important ones iteratively.
  - **Stepwise selection (forward, backward, or mixed):** Adds or removes predictors based on information criteria like AIC or BIC.
  - **Regularization methods (e.g., Lasso):** Simultaneously penalize over-complexity and encourage sparse solutions by shrinking coefficients of irrelevant features toward zero.
  - **Filter methods** (like SVM-based or information-gain): Evaluate each feature’s predictive value individually or in combinations, before model fitting.

Recent studies show that **hybrid approaches** combining automated coarse tuning with manual fine-tuning yield robust, interpretable models without excessive computational cost. Automated techniques efficiently screen large sets of variables, while manual review ensures scientific and practical relevance in the final model.

### Recursive Feature Elimination

Recursive Feature Elimination (RFE) is a widely used and effective feature selection method in machine learning that operates by sequentially removing features and assessing model performance to identify the most important predictors. Unlike simple backward elimination, RFE uses model-based feature importance rankings to guide the elimination process, requiring fewer trained models and reducing computational cost.

These notes cover Recursive Feature Elimination (RFE), a wrapper technique for automated feature selection based on model-derived feature importance or rankings. RFE iteratively prunes the least important features, balancing model performance and simplicity.

RFE trains a machine learning model on all features, ranks them by importance (e.g., coefficients or importances), and recursively eliminates the weakest until a target number is reached or performance drops.

1. **Initial model training:**  
   Build a machine learning model (e.g., random forest, gradient boosting, logistic regression) using all available features.

2. **Feature ranking:**  
   Rank features according to their importance scores or coefficients derived from the trained model.

3. **Feature removal:**  
   Remove the least important feature from the dataset.

4. **Retrain the model and evaluate:**  
   Train a new model with the reduced set of features and evaluate its performance.

5. **Performance comparison:**  
   Compare current model performance with the previous step. If performance decreases beyond a pre-defined threshold, reconsider removal (sometimes keep the feature).

6. **Iteration:**  
   Repeat steps 3 to 5 until all features have been assessed or a stopping criterion is met.

**Practical Benefits of RFE:**

- Effectively **reduces feature dimensionality**, improves model interpretability, and decreases training time.  
- Helps mitigate **overfitting** by removing irrelevant or redundant variables.  
- Particularly useful in high-dimensional contexts, such as bioinformatics, medical signal processing, and text classification.


**Stopping Criteria**: Manual (`n_features_to_select`) or automatic (RFECV monitors CV score drop).

  | Attribute          | Description |
  |--------------------|-------------|
  | `support_`        | Boolean mask of kept features |
  | `ranking_`        | Elimination order (1 = top-selected) |
  | `n_features_`     | Final count (RFECV-optimized) |
  | `grid_scores_`    | CV scores per feature count (RFECV) |

## Regularization

To know more about it, Read '[How regularization works conceptually?](https://explained.ai/regularization/constraints.html)' article

- **Linear and logistic regression models** are fundamental in machine learning because they are highly **interpretable**, computationally **fast**, and serve as the foundation for more complex models like deep neural networks.  
- Despite their simplicity of fitting lines (or hyperplanes) to training data, linear models often fit too closely to outliers, leading to poor generalization on new data.  
- To improve generalization, **regularization** techniques are commonly employed—these impose constraints on the model parameters to prevent overfitting.  
- While there are many forms of regularization (e.g., early stopping, dropout for deep learning), for isolated linear models, **Lasso (L1)** and **Ridge (L2)** are the most widely used.

### Why Regularization?

- Overfitted models capture noise as if it were signal, reducing predictive quality on new data.  
- Regularization adds a **penalty term** to the loss function to control model complexity while fitting training data, improving robustness and generalization.
- Applying regularization has the effect of reducing the slope w and increasing the intercept. This produces a regression line that is slightly insensitive to
`X` and not too sensitive.

  
### Cost Function with Regularization

- Ordinary least squares (OLS) minimizes Residual Sum of Squares (RSS):  
$$
\text{RSS} = \sum_{i=1}^n (y_i - \hat{y}_i)^2
$$

- Regularized cost function adds a penalty proportional to the coefficients:  
$$
\text{Cost} = \text{RSS} + \lambda \times \text{Reg}(\mathbf{w})
$$

### How Model Complexity Increases

- Complexity can increase by:  
  1. Large magnitude coefficients  
  2. Large number of features (nonzero coefficients)  

- Regularization penalizes these through the term $\text{Reg}(w)$ that discourages complex models.
- **Normalization or Standardization** is necessary, especially when using regularization. Without normalization, **the regularization term may unfairly impose a greater
penalty on some coefficients than on others**.

### Types of Regularization Techniques

#### Ridge Regression (L2)

- Penalty: sum of squares of coefficients ($\sum w_j^2$)  
- Geometrically, the coefficients are restricted inside a **circle** (2D: $\beta_0^2 + \beta_1^2 \leq c$)  
- Coefficients shrink towards zero but none become zero—**no feature elimination**  
- Handles multicollinearity by stabilizing coefficients  
- Has analytical or fast computational solutions

- Requires iterative solvers, computationally more intensive  
- Produces sparse, interpretable models

### Geometric  Intuition

- Optimal coefficients come from where Residual Sum of Squares (RSS) contours tangent to the constraint region.  
- Ridge’s smooth circular constraint results in coefficients rarely being exactly zero.  
- Lasso’s corners increase likelihood coefficients hit zero — performs feature selection.  
- Increasing $\lambda$ shrinks constraint size, moving the coefficients closer to zero (more regularization).

### Effect of $\lambda $ on Bias-Variance Tradeoff

- $\lambda = 0$: Coefficients are OLS estimates, model has minimum bias but high variance.  
- Increasing $\lambda$: Coefficients shrink, bias increases but variance decreases.  
- $\lambda \to \infty$: All coefficients are zero; model predicts mean; maximum bias, minimum variance.

- Regularized regression models are crucial for creating **robust, generalizable** predictive models.  
- They provide a foundation for tackling high-dimensional data and avoiding overfitting common in complex or real-world datasets.
